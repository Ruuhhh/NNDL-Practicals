```python
import numpy as np

# Part-1 Vector addition
arr1 = np.array([2,3,4])
arr2 = np.array([4,6,8])

print("First array is : ",arr1)
print("Second array is : ",arr2)

add = arr1 + arr2

print("Vector addition is: ", add)
```

```python
# Part-2 Vector Multiplication

mat1 = np.array([[2,3,4],[4,5,6],[1,4,5]])
mat2 = np.array([[3,3,4],[5,5,6],[4,4,5]])

print("First array is : \n",mat1)
print("\nSecond array is : \n",mat2)

mul = np.matmul(mat1,mat2)

print("\nMatrix multiplication is: \n", mul)
```
